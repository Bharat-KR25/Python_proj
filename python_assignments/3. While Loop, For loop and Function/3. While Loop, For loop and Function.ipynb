```python
import random
secret_number= random.randint(1, 10)
attempts =1
i=1
while attempts <=3:
  i= int(input("enter the number"))
  if i<=10 :
    print(f"Attempt no {attempts}/3")
    if i == secret_number:
       print("You Won")
       break
    else :
      print("Try Again")
      attempts+=1
      continue
  else :
    print("Warning, enter values from 1 to 10")
else :
  print("Attempts over, Game over")
  
```

```python
def calculate_bmi(Weight, Height):
  Height_m = Height * 0.3048
  BMI = (Weight/Height_m**2)
  return BMI
    
Weight = float(input("Enter your Weight in(kgs)"))
Height = float(input("Enter your Height in (ft)"))
Your_BMI = calculate_bmi(Weight, Height)

print(f"Your BMI= {round(Your_BMI,2)}")
```
